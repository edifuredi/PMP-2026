# Problema Bayes – test de diagnostic

Se consideră o boală \(B\) care afectează 1% din populație:

\[
P(B)=0.01
\]

Testul are:

- sensibilitate de 95%:

\[
P(+|B)=0.95
\]

- specificitate de 90%:

\[
P(-|\overline{B})=0.90
\]

Prin urmare:

\[
P(+|\overline{B})=1-0.90=0.10
\]


## a) Probabilitatea ca o persoană să fie bolnavă dacă testul este pozitiv

Folosim teorema lui Bayes:

\[
P(B|+)=
\frac{P(+|B)\cdot P(B)}
{P(+|B)\cdot P(B)+P(+|\overline{B})\cdot P(\overline{B})}
\]

\[
P(\overline{B})=1-0.01=0.99
\]

\[
P(B|+)=
\frac{0.95\cdot0.01}
{0.95\cdot0.01+0.10\cdot0.99}
\]

\[
P(B|+)=
\frac{0.0095}{0.1085}
\]

\[
P(B|+)\approx0.0876
\]

\[
\boxed{P(B|+)\approx8.76\%}
\]


## b) Modelare în Python

Simulăm un număr mare de persoane și estimăm probabilitatea ca o persoană să fie bolnavă în condițiile în care testul este pozitiv.


```python
import random

N = 100000

bolnavi_pozitivi = 0
total_pozitivi = 0

for i in range(N):
    bolnav = random.random() < 0.01

    if bolnav:
        pozitiv = random.random() < 0.95
    else:
        pozitiv = random.random() < 0.10

    if pozitiv:
        total_pozitivi += 1

        if bolnav:
            bolnavi_pozitivi += 1

probabilitate = bolnavi_pozitivi / total_pozitivi

print("Numar teste pozitive:", total_pozitivi)
print("Bolnavi cu test pozitiv:", bolnavi_pozitivi)
print("P(B|Pozitiv) =", probabilitate)
print("P(B|Pozitiv) =", probabilitate * 100, "%")
```

## c) Specificitatea minimă pentru ca probabilitatea să ajungă la 50%

Dorim:

\[
P(B|+)=0.5
\]

Notăm specificitatea cu \(s\). Atunci:

\[
P(+|\overline{B})=1-s
\]

Aplicăm formula lui Bayes:

\[
0.5=
\frac{0.95\cdot0.01}
{0.95\cdot0.01+(1-s)\cdot0.99}
\]

\[
0.5=
\frac{0.0095}
{0.0095+0.99(1-s)}
\]

\[
0.5[0.0095+0.99(1-s)]=0.0095
\]

\[
0.0095+0.99(1-s)=0.019
\]

\[
0.99(1-s)=0.0095
\]

\[
1-s=\frac{0.0095}{0.99}
\]

\[
s\approx0.990404
\]

\[
\boxed{s\approx99.04\%}
\]
